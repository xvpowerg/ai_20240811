## 單位相同的比較

```python
m1 = 60
s1 = 20

m2 = 70
s2 = 10

a = 90
b = 90
r1 = (a - m1) / s1
r2 =  (b - m2) / s2
print(r1,r2)
if r1 > r2:
    print("A同學成績好")
elif r1 < r2:
    print("B同學成績好")
else:
    print("同學成績一樣好")
```

```python
import numpy as np
a1 = np.array([[6,3],
               [8,10]])
b1 = np.array([[2,1],
               [5,-3]])
print(a1 + b1)

print(a1 - b1)
```

```python
import numpy as np
a = [[15,14,15],
     [12,13,17]]
b = [[10,14],
     [5,18],
     [8,10]]
a = np.array(a)
b = np.array(b)
print(np.dot(a,b))
print(a @ b)
```

```python
import numpy as np
a = [[6,3],
     [8,10]]
b = [[2,1],
     [5,-3]]
a = np.array(a)
b = np.array(b)
print(a @ b)
print(b @ a)
```

```python
import numpy as np
a = np.array([[4,-7],
             [2,-3]])
a_inv = np.linalg.inv(a)
print(a_inv)
print(a_inv @ a)
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(1,8)
y = [1,3,3,12,5,7,9]
plt.plot(x,y)
plt.show()
```

```python
import numpy as np
slope,intercept = np.polyfit(x,y,1)
# y = mx + b
y1 = [slope*i + intercept for i in x]
plt.plot(x,y,'--')
plt.plot(x,y1,'b')
plt.show()
```

```python
import numpy as np
slope1,slope2,intercept =  np.polyfit(x,y,2)
print(slope1,slope2,intercept)
y1 = [slope1*i**2 + slope2*i + intercept for i in x]
plt.plot(x,y,'--')
plt.plot(x,y1,'b')
```

## 聯立方程式計算

```python
import sympy as sp
x = sp.Symbol("x")
y = sp.Symbol("y")
f1 = 2*x + 3*y - 23
f2 = x + 4 * y - 24
ans = sp.solve([f1,f2],[x,y])
print(ans)

for k in ans:
    print(k,ans[k])
print(ans[x])    
print(ans[y])    
```

```python
import sympy as sp
x = sp.Symbol("x")
y = sp.Symbol("y")
f1 = x + y - 20
f2 = 50 * x + 30 * y - 800
print(sp.solve([f1,f2],[x,y]))
```

```python
import numpy as np
a1 = np.array([[1,50],
              [1,30]])
a2 = np.array([20,800])
a1_inv = np.linalg.inv(a1)
print(a2 @ a1_inv)
```

```python
import pandas as pd
tvMarketing = pd.read_csv("tvmarketing.csv")
print(tvMarketing.head())
print(tvMarketing.isnull().any())
print(tvMarketing.dtypes)
```

```python
import matplotlib.pyplot as plt
plt.scatter(tvMarketing['TV'],tvMarketing['Sales'])
plt.xlabel('TV')#營銷預算
plt.ylabel('Sales')#銷售額
plt.show()
```

```python
from sklearn.model_selection import train_test_split
import sklearn.linear_model
print('我們先建立X與y資料，後續才能進行切割')
X = tvMarketing["TV"]
y = tvMarketing["Sales"]
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.3)
print(x_train.shape)
print(x_test.shape)
```

```python
model = sklearn.linear_model.LinearRegression()
print(x_train.shape)
x_train = np.array(x_train).reshape(-1,1)
print(x_train.shape)
model.fit(x_train,y_train)
print(model.coef_)
print(model.intercept_)
```

```python
import matplotlib.pyplot as plt
y2 = model.coef_ * X + model.intercept_
plt.scatter(X,y)
plt.scatter(X,y2,c="red")
plt.show()
```
